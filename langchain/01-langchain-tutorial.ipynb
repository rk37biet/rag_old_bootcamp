## Create a Chatbot

```python
chat_history = [
    SystemMessage(content="You are a helpful assistant")
]
```

```python
while True:
    user_input=input("user_input: ")
    chat_history.append(HumanMessage(content=user_input))
    if user_input=="exit":
        break
    result=openai_model.invoke(chat_history)
    chat_history.append(AIMessage(content=result.content))
    print("AI Generate Answer: ", result.content)
```

```python
print("Chat History: ", chat_history)
```

### Sequential LLM Chain

```python
prompt1 = PromptTemplate(
    template="get a detailed report on {topic}",
    input_variables=["topic"]
)

prompt2 = PromptTemplate(
    template="generate a 3 point of summary from the following {text}",
    input_variables=["text"]
)
```

```python
# Sequential LLM Chain
sequential_chain = prompt1 | openai_model | parser | prompt2 | openai_model | parser
```

```python
sequential_chain.get_graph().print_ascii()
```

```python
result = sequential_chain.invoke({"topic":"machine learning"})
```

### Parallel Chaining using LCEL

```python
prompt1 = PromptTemplate(
    template="generate simple summary from the following text \n {text}",
    input_variables=["text"]
)

prompt2 = PromptTemplate(
    template="generate 2 question and answer pairs from the following text \n {text}",
    input_variables=["text"]
)

prompt3 = PromptTemplate(
    template="merge the provided summary and question, answer pairs into a single document \n summary: {summary}, \n Q&A: {qa}",
    input_variables=["summary","qa"]
)
```

```python
from langchain.schema.runnable import RunnableParallel
```

```python
# Parallel LLM Chain using RunnableParallel in LCEL
parallel_chain = RunnableParallel({
    "summary": prompt1 | openai_model | parser,
    "qa": prompt2 | openai_model | parser
}
)
```

```python
merge_chain = prompt3 | openai_model | parser
```

```python
chain = parallel_chain | merge_chain
```

```python
chain.get_graph().print_ascii()
```
