## Langchain Wrapper for LLM Model Integration

```python
# langchain wrapper for llm model integration
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
```

### Using HuggingFace Chat Model

```python
# load hugginface model using HugginFaceEndpoint wrapper
llm = HuggingFaceEndpoint(repo_id="HuggingFaceH4/zephyr-7b-beta",task="text-generation")
```

```python
# invoke llm using ChatHuggingFace wrapper
# we are invoking hugginface modle directly from API, not downloading the model locally
model = ChatHuggingFace(llm=llm,max_new_tokens=256)
```

```python
model.invoke("What is the capital of USA?").content
```

## Embedding Models
- Opensource embedding model ex- huggingface
- OpenAI ex- text-embedding

```python
from langchain_openai import OpenAIEmbeddings
```

```python
# openai large embedding model
openai_embedding = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
# generate embedding for query
result = openai_embedding.embed_query("India is a growing country")
```

```python
# dim of embedding vector
len(result)
```

```python
#openai small embedding model
openai_embedding_small = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
result2= openai_embedding_small.embed_query("India is a growing country")
```

```python
len(result2)
```

```python
# create embedding vector of size 64 using openai text-embedding-3-large model
openai_embedding_large_64 = OpenAIEmbeddings(model="text-embedding-3-large",dimensions=64)
```

```python
result_large_64 = openai_embedding_large_64.embed_query("India is a growing country")
```

```python
len(result_large_64)
```

### Embed documents

```python
documents = ["what is a capital of USA",
             "who is a president of usa",
             "who is a prime minister of india"]
```

```python
doc_embeddings=openai_embedding_large_64.embed_documents(documents)
```

```python
len(doc_embeddings)
```

```python
doc_embeddings
```

```python
len(doc_embeddings[0])
```

## Simialrity Search
- my_query is more similar to which document in the list of documents?

```python
documents = ["what is a capital of USA",
             "who is a president of usa",
             "who is a prime minister of india"]
             
my_query = "narendra modi is indian prime minister"
```

```python
query_embedding = openai_embedding_large_64.embed_query(my_query)
```

```python
query_embedding
```

```python
len(query_embedding)
```

```python
# similarity search using cosine similarity
from sklearn.metrics.pairwise import cosine_similarity
```

```python
scores=cosine_similarity([query_embedding], doc_embeddings)
```

```python
# my_query is more similar to "who is a prime minister of india" document as compared to other 2 documents based on cosine similarity scores
scores
```
